```python
import ttkbootstrap as ttk


# Fenster erstellen
app = ttk.Window(themename="flatly")
app.title("Minimales Treeview-Beispiel")
app.geometry("500x250")

# Spalten definieren
columns = ("name", "alter", "stadt")

# Tabelle erstellen
tree = ttk.Treeview(
    app,
    columns=columns,
    show="headings"
)

# Spaltenüberschriften
tree.heading("name", text="Name")
tree.heading("alter", text="Alter")
tree.heading("stadt", text="Stadt")

# Spaltenbreiten
tree.column("name", width=180)
tree.column("alter", width=80, anchor="center")
tree.column("stadt", width=180)

# Beispiel-Zeilen einfügen
tree.insert("", "end", values=("Max Mustermann", 32, "Berlin"))
tree.insert("", "end", values=("Erika Musterfrau", 28, "Hamburg"))
tree.insert("", "end", values=("Paul Beispiel", 45, "München"))

# Tabelle anzeigen
tree.pack(fill="both", expand=True, padx=10, pady=10)

app.mainloop()
```
